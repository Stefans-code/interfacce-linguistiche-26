### Variabili
Una variabile è un contenitore che memorizza un valore.

**tipi principali**

### Operatori principali

Aritmetici: `+` `-` `*` `/` `//` (divisione intera) `%` (resto) `**` (potenza)

### Stringhe
Testo racchiuso tra virgolette.
Supportano molte operazioni e sono un **iterabile**, quindi possiamo prenderne soltanto dei "pezzetti" (slice).



```python
la_mia_stringa = "pippo"
stringa_vuota = ""
```

```python
print(len(la_mia_stringa))
print(len(stringa_vuota))
```

```python
s = "Python"
print(len(s))         # lunghezza
print(s.upper())      # maiuscolo
print(s.lower())      # minuscolo
print(s[0])           # primo carattere (P)
print(s[-1])          # ultimo carattere (n)
print(s.replace("Py", "My"))  # sostituzione
print(s[0:3]) # primi 3 caratteri (cioè i caratteri 0, 1, 2). l'indice 3 non è incluso.
```

```python
s.index('y') # questo mi dice a che indice si trova 'y' in s (che per noi corrisponde a 'Python')
```

```python
print(s.count('t')) # quante t ci sono in s?
print('tutto'.count('t')) # quante t ci sono in 'tutto'?
```

```python
# split è un'altra funzione che ci tornerà molto utile
stringa_con_spazi = "ciao questa stringa contiene diversi spazi"
parole = stringa_con_spazi.split(" ") # qui dico a python di dividere la stringa ogni volta che trova uno spazio
print(parole) # questa è una lista, a breve vediamo cosa sono le liste
```


### Funzioni
Una funzione è un blocco di codice riutilizzabile.
Si definisce con `def`.

Le funzioni di solito hanno dei **parametri**, cioè valori che la funzione riceve
**Return**: il risultato che la funzione restituisce

```python
def saluta(nome):
    return f"Ciao {nome}!"

print(saluta("Camilla"))
```

### Note sull'indentazione in Python

1. Cos’è l’indentazione

    L’indentazione sono spazi o tab all’inizio di una riga.

    Serve a raggruppare le istruzioni che appartengono a uno stesso blocco (per esempio il corpo di un if, di un for, di una funzione, ecc.).

    Python non ignora gli spazi come molti altri linguaggi: un’indentazione sbagliata genera un errore (IndentationError).

2. Quanto indentare

    Convenzione comune: 4 spazi per livello di indentazione. Tipicamente un blocco lo troviamo dopo i due punti (`:`)

    Non mischiare tab e spazi nello stesso progetto, perché Python lo vede come errore.

### `if`, `elif`, `else`
- `if`: controlla se una condizione è vera. Se sì, esegue il blocco di codice indentato sotto.

- `elif` (else if): permette di controllare altre condizioni, se la precedente era falsa.

- `else`: viene eseguito se tutte le condizioni precedenti sono false.

```python
# Esempio base
x = 7
if x > 10:
    print("x è maggiore di 10")
elif x > 5:
    print("x è maggiore di 5 ma minore o uguale a 10")
else:
    print("x è minore o uguale a 5")
```

```python
# Esempio con stringhe
nome = "Anna"
if nome == "Camilla":
    print("Ciao Camilla!")
elif nome == "Anna":
    print("Ciao Anna!")
else:
    print("Ciao, chiunque tu sia!")
```

### Liste
Le liste sono come gli array. Permettono di memorizzare una sequenza ordinata di elementi. Gli elementi possono essere di qualsiasi tipo (numeri, stringhe, altre liste, ecc.).


Le liste sono **iterabili**,
quindi le possiamo spezzettare come le stringhe o recuperare solo un elemento alla volta.


- Le liste sono **mutabili**: posso modificarne il contenuto.
- Gli elementi sono indicizzati: il primo elemento ha indice 0.
- Posso accedere agli elementi con la sintassi `lista[indice]`.


```python
numeri = [1, 2, 3, 4, 5]
nomi = ["Anna", "Luca", "Camilla"]
mista = [1, "ciao", 3.14, True]
```

```python
print(numeri[0])      # Stampa 1
print(nomi[2])        # Stampa "Camilla"
```

```python
# modificare una lista
numeri[0] = 10        # Ora numeri diventa [10, 2, 3, 4, 5]
```

#### Funzioni e metodi utili
- `len(lista)` → restituisce la lunghezza della lista

- `lista.append(x)` → aggiunge l’elemento x alla fine

- `lista.insert(i, x)` → inserisce x in posizione i

- `lista.remove(x)` → rimuove la prima occorrenza di x

- `lista.pop(i)` → rimuove e restituisce l’elemento in posizione i

- `lista.sort()` → ordina la lista (solo se gli elementi sono confrontabili)

- `lista.reverse()` → inverte l’ordine degli elementi

```python
frutti = ["mela", "banana", "kiwi"]
frutti.append("arancia")      # ["mela", "banana", "kiwi", "arancia"]
frutti.remove("banana")       # ["mela", "kiwi", "arancia"]
print(len(frutti))
print(frutti)# 3
```

### Slicing
Come con le stringhe, posso estrarre una parte della lista usando la sintassi lista[start:stop], in cui *start* è incluso, mentre *stop* è escluso

```python
numeri = [1, 2, 3, 4, 5]
```

```python
print(numeri[1:4])
```

```python
print(numeri[:3])
```

```python
print(numeri[2:])
```

### Cicli `for`

Un ciclo `for` è utilizzato per iterare su una sequenza (come una lista, una stringa o un range di numeri).


```python
parola = "Python"

for lettera in parola:
    print(lettera)
```

```python
frutti = ["mela", "banana", "kiwi"]

for frutto in frutti:
    print(frutto)
```

```python
# range()
# La funzione range() genera una sequenza di numeri. Posso specificare inizio, fine (escluso) e passo.

# Da 0 a 4
for i in range(5):
    print(i)
```

```python
# Da 1 a 5
for i in range(1, 6):
    print(i)
```

```python
# Da 0 a 10, passo 2
for i in range(0, 11, 2):
    print(i)
```

```python
nomi = ["Anna", "Luca", "Camilla"]

for i in range(len(nomi)):
    print(f"L'indice {i} contiene: {nomi[i]}")
```
